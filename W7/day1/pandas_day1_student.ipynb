# Part 2: Understanding Data Types

## Duration: ~25 minutes

## The Two Fundamental Types of Data

```
ALL DATA IN THE WORLD FALLS INTO TWO CATEGORIES:

    ┌────────────────────────────────────────────────────────────────────┐
    │                                                                    │
    │   QUALITATIVE (Categorical)      QUANTITATIVE (Numerical)        │
    │   ─────────────────────────      ──────────────────────────       │
    │                                                                    │
    │   Describes QUALITIES            Describes QUANTITIES             │
    │   "What kind?" "Which one?"       "How much?" "How many?"          │
    │                                                                    │
    │   Categories / Labels            Numbers you can calculate        │
    │   Can COUNT them                 Can do MATH on them              │
    │                                                                    │
    │   Examples:                      Examples:                        │
    │   • Gender (Male/Female)         • Age (25, 30, 45 years)         │
    │   • Country (USA, Israel, UK)    • Height (175cm, 180cm)          │
    │   • Product type (Book, Toy)     • Price ($10.99, $24.99)         │
    │   • Color (Red, Blue, Green)     • Temperature (22°C, 35°C)       │
    │   • Yes/No answers               • Weight (70kg, 85kg)            │
    │                                                                    │
    └────────────────────────────────────────────────────────────────────┘
    
    
THE KEY QUESTION:
    
    "Can I calculate an average of this?"
    
    YES → Quantitative
    NO  → Qualitative
```

## Why Data Types Matter: A Real Example

```
SCENARIO: Analyzing customer satisfaction survey

    Your data looks like this:
    ┌──────────┬────────┬─────────────┬────────────┐
    │ Customer │ Rating │ Country     │ Age        │
    ├──────────┼────────┼─────────────┼────────────┤
    │ C001     │ "5"    │ USA         │ 34         │
    │ C002     │ "3"    │ UK          │ 28         │
    │ C003     │ "4"    │ USA         │ 45         │
    └──────────┴────────┴─────────────┴────────────┘
    
    
    PROBLEM: Rating is stored as TEXT ("5" not 5)
    
    You try: average_rating = sum(ratings) / len(ratings)
    
    Result:  ERROR! "TypeError: unsupported operand type for +: 'str'"
             (Translation: You can't add text together!)
    
    
    SOLUTION: Convert "5" → 5 (text to number)
    
    Then:    average_rating = 4.0 ✓
```

**Lesson:** Always check your data types before analysis!

## Tricky Cases: Numbers That Are Actually Categories

```
COMMON TRAPS - These LOOK like numbers but are CATEGORICAL:

    ┌─────────────────────────────────────────────────────────────────────┐
    │                                                                     │
    │   ⚠️  ZIP CODES: 10001, 90210, 60601                               │
    │       They're digits, but... average zip code = meaningless!       │
    │       These are LABELS for locations, not quantities.              │
    │                                                                     │
    │   ⚠️  PHONE NUMBERS: 0521234567, 2125551234                        │
    │       Digits, but adding phone numbers makes no sense.             │
    │       These are IDENTIFIERS, not quantities.                       │
    │                                                                     │
    │   ⚠️  ID NUMBERS: Employee ID, Student ID, Order ID                │
    │       1001, 1002, 1003... but they're just labels!                 │
    │       Average employee ID = completely meaningless.                │
    │                                                                     │
    │   ⚠️  YEARS AS CATEGORIES: Survey year 2020, 2021, 2022            │
    │       Sometimes you calculate with years (age = 2025 - birth_year) │
    │       Sometimes they're categories (compare 2020 vs 2021 data)     │
    │       → Depends on your analysis!                                  │
    │                                                                     │
    │   ⚠️  JERSEY NUMBERS: Player #23, #7, #10                          │
    │       Is #23 "greater than" #7? No! They're labels.               │
    │                                                                     │
    └─────────────────────────────────────────────────────────────────────┘
    
THE TEST: "Does it make sense to ADD these numbers or find their AVERAGE?"
    
    NO → It's categorical (qualitative), even if it looks like numbers!
```

## Qualitative Data: Two Subtypes

```
QUALITATIVE DATA HAS TWO SUBTYPES:

    ┌─────────────────────────────────────────────────────────────────────┐
    │                                                                     │
    │   NOMINAL (Named)                 ORDINAL (Ordered)                │
    │   ─────────────────              ─────────────────                 │
    │                                                                     │
    │   NO natural order               HAS natural order                 │
    │   Just different categories      Categories have ranking           │
    │                                                                     │
    │   Examples:                      Examples:                         │
    │   • Colors (Red, Blue, Green)    • Education (HS → BS → MS → PhD) │
    │   • Countries (USA, UK, Israel)  • Size (XS → S → M → L → XL)     │
    │   • Blood type (A, B, AB, O)     • Star rating (1 → 2 → 3 → 4 → 5)│
    │   • Gender (M, F, Other)         • Satisfaction (Low → Med → High)│
    │   • Browser (Chrome, Firefox)    • Income level (Low → Mid → High)│
    │                                                                     │
    │   TEST: Can you say one is       TEST: Can you rank them?          │
    │   "greater" than another?        Is PhD > BS? Yes!                 │
    │   Is Red > Blue? NO!             Is XL > S? Yes!                   │
    │                                                                     │
    └─────────────────────────────────────────────────────────────────────┘
```

## Quantitative Data: Two Subtypes

```
QUANTITATIVE DATA HAS TWO SUBTYPES:

    ┌─────────────────────────────────────────────────────────────────────┐
    │                                                                     │
    │   DISCRETE (Countable)            CONTINUOUS (Measurable)          │
    │   ────────────────────            ───────────────────────          │
    │                                                                     │
    │   Whole numbers only              Any value in a range             │
    │   You COUNT them                  You MEASURE them                 │
    │   Finite possible values          Infinite possible values         │
    │                                                                     │
    │   Examples:                       Examples:                        │
    │   • Number of children (0,1,2,3)  • Height (175.5 cm, 175.52 cm)   │
    │   • Items purchased (1, 2, 3)     • Weight (68.3 kg, 68.37 kg)     │
    │   • Website visits (100, 101)     • Temperature (23.7°C)           │
    │   • Eggs in a carton (0-12)       • Time duration (3.5 hours)      │
    │   • Goals scored (0, 1, 2, 3)     • Distance (5.234 km)            │
    │                                                                     │
    │   TEST:                           TEST:                            │
    │   Can you have 2.5 of it?         Can it be any decimal?           │
    │   2.5 children? NO!               175.5234 cm tall? YES!           │
    │   → Discrete                      → Continuous                     │
    │                                                                     │
    └─────────────────────────────────────────────────────────────────────┘
```

## Data Types: The Complete Picture

```
DATA TYPES DECISION TREE:

                            DATA
                             │
              ┌──────────────┴──────────────┐
              │                             │
              ▼                             ▼
        QUALITATIVE                   QUANTITATIVE
        (Categories)                    (Numbers)
              │                             │
       ┌──────┴──────┐               ┌──────┴──────┐
       │             │               │             │
       ▼             ▼               ▼             ▼
    NOMINAL      ORDINAL         DISCRETE    CONTINUOUS
   (No order)   (Ordered)       (Counted)    (Measured)
       │             │               │             │
       ▼             ▼               ▼             ▼
    Colors       Ratings        # of items     Height
    Countries    Sizes          # of people    Weight
    Gender       Education      # of clicks    Temperature


QUICK DECISION FLOW:

    1. Can I calculate an average?
       NO  → Qualitative → Can I rank them? → Yes=Ordinal, No=Nominal
       YES → Quantitative → Must it be whole numbers? → Yes=Discrete, No=Continuous
```

## Part 2 Summary: Data Types Cheat Sheet

```
QUICK REFERENCE:

    TYPE          │ MEANING              │ EXAMPLES                    │ OPERATIONS
    ──────────────┼──────────────────────┼─────────────────────────────┼─────────────────
    Nominal       │ Categories, no order │ Colors, Countries           │ Count, Mode
    Ordinal       │ Categories, ordered  │ Sizes, Ratings              │ Count, Mode, Median
    Discrete      │ Whole numbers        │ # of items, # of people     │ All math operations
    Continuous    │ Any number           │ Height, Weight, Temperature │ All math operations


COMMON MISTAKES TO AVOID:

    ❌ Calculating average of zip codes
    ❌ Treating IDs as quantities
    ❌ Not converting string numbers to actual numbers
    ❌ Treating star ratings as purely numerical


REMEMBER:

    "Just because it's made of digits doesn't make it quantitative!"
```

## Loading Data: pd.read_csv()

```
THE MOST IMPORTANT FUNCTION IN PANDAS:

    df = pd.read_csv("filename.csv")
    
    • Reads a CSV file into a DataFrame
    • CSV = "Comma Separated Values" (spreadsheet as text)
    • Returns a DataFrame object
    • "df" is the conventional variable name for DataFrames


WHAT'S A CSV FILE?

    A CSV is just text:
    
    Name,Age,City,Salary        ← Header row
    Alice,28,New York,75000     ← Data row 1
    Bob,35,Chicago,82000        ← Data row 2
    Charlie,42,Miami,91000      ← Data row 3
    
    Values separated by commas. That's it.
    Excel can export to CSV. Google Sheets can export to CSV.
    It's the universal data exchange format.


OTHER LOADING FUNCTIONS:

    pd.read_excel("file.xlsx")    # Excel files
    pd.read_json("file.json")     # JSON files
    pd.read_sql(query, connection) # SQL databases
    pd.read_html(url)              # Tables from web pages!
```

## The Iris Dataset: A Classic

```
THE IRIS DATASET (1936)

    Created by: Ronald Fisher, British statistician
    
    Contains: Measurements of 150 iris flowers
              3 species × 50 flowers each
    
    Columns:
    • sepal_length (cm)  ← Length of outer petals
    • sepal_width (cm)   ← Width of outer petals
    • petal_length (cm)  ← Length of inner petals
    • petal_width (cm)   ← Width of inner petals
    • species            ← setosa, versicolor, virginica
    
              Sepal
               ↓
        ╭─────────────╮
       ╱   ╭─────╮    ╲
      │   │ Petal│    │
       ╲   ╰─────╯    ╱
        ╰─────────────╯
    
    Why it's famous: Simple, clean, perfect for learning.
    Every data science tutorial uses it!
```

````python
# Load the Iris dataset from sklearn (it's built-in)
from sklearn.datasets import load_iris

# Load the raw data
iris_data = load_iris()

# Create a DataFrame
df = pd.DataFrame(
    data=iris_data.data,
    columns=['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
)

# Add the species column
# Convert numbers (0,1,2) to names (setosa, versicolor, virginica)
df['species'] = [iris_data.target_names[i] for i in iris_data.target]

# Show that it worked
print(f"Dataset loaded successfully!")
print(f"Shape: {df.shape[0]} rows × {df.shape[1]} columns")
````

# Part 5: Basic Statistics

## Mean, Median, and Mode

## Duration: ~25 minutes

## The Three M's: Mean, Median, Mode

```
MEASURES OF CENTRAL TENDENCY:

    "What is the TYPICAL value in this data?"
    
    ┌─────────────────────────────────────────────────────────────────────┐
    │                                                                     │
    │   MEAN (Average)                                                   │
    │   ─────────────────                                                │
    │   Sum of all values ÷ Count of values                              │
    │                                                                     │
    │   Example: [2, 4, 6, 8, 10]                                        │
    │   Mean = (2+4+6+8+10) / 5 = 30/5 = 6                               │
    │                                                                     │
    │   Use when: Data is numeric, no extreme outliers                   │
    │                                                                     │
    ├─────────────────────────────────────────────────────────────────────┤
    │                                                                     │
    │   MEDIAN (Middle)                                                  │
    │   ─────────────────                                                │
    │   The middle value when sorted                                     │
    │                                                                     │
    │   Example: [2, 4, 6, 8, 10]                                        │
    │   Sorted → [2, 4, 6, 8, 10] → Middle = 6                           │
    │                                                                     │
    │   Use when: Data has outliers (income, house prices)               │
    │                                                                     │
    ├─────────────────────────────────────────────────────────────────────┤
    │                                                                     │
    │   MODE (Most Common)                                               │
    │   ─────────────────                                                │
    │   The value that appears most often                                │
    │                                                                     │
    │   Example: [2, 4, 4, 6, 8]                                         │
    │   Mode = 4 (appears twice, others appear once)                     │
    │                                                                     │
    │   Use when: Data is categorical, or finding most common value     │
    │                                                                     │
    └─────────────────────────────────────────────────────────────────────┘
```

## Why Three Different Measures?

```
WHEN MEAN AND MEDIAN DIFFER: THE OUTLIER PROBLEM

    Consider salaries at a small company:
    
    Employee 1:    $50,000
    Employee 2:    $55,000
    Employee 3:    $60,000
    Employee 4:    $65,000
    CEO:        $1,000,000
    
    MEAN salary:   ($50k + $55k + $60k + $65k + $1M) / 5 = $246,000
    MEDIAN salary: $60,000 (the middle value)
    
    
    Which better represents a "typical" salary?
    
    The MEAN is pulled up by the CEO's salary.
    The MEDIAN tells you what the middle person actually earns.
    
    
    ┌─────────────────────────────────────────────────────────────────────┐
    │                                                                     │
    │   RULE OF THUMB:                                                   │
    │                                                                     │
    │   • Mean: Best for "normal" data without extreme values           │
    │   • Median: Best when outliers exist (income, house prices)       │
    │   • Mode: Best for categorical data or finding most common        │
    │                                                                     │
    │   "Median household income" - not "mean" - because of billionaires│
    │                                                                     │
    └─────────────────────────────────────────────────────────────────────┘
```

````python
# Let's calculate Mean, Median, Mode with our Iris data
print("=" * 60)
print("STATISTICS FOR SEPAL LENGTH")
print("=" * 60)

# Mean
mean_val = df['sepal_length'].mean()
print(f"\nMEAN (average):  {mean_val:.2f} cm")

# Median
median_val = df['sepal_length'].median()
print(f"MEDIAN (middle): {median_val:.2f} cm")

# Mode
mode_val = df['sepal_length'].mode()[0]  # mode() returns a Series
print(f"MODE (most common): {mode_val:.2f} cm")

print("\n" + "=" * 60)
print("INTERPRETATION")
print("=" * 60)
print(f"\nThe average sepal length is {mean_val:.2f} cm.")
print(f"Half of flowers have sepal length below {median_val:.2f} cm.")
print(f"The most common sepal length is {mode_val:.2f} cm.")
````

````python
# Mode is especially useful for categorical data!
print("=" * 60)
print("MODE FOR CATEGORICAL DATA")
print("=" * 60)

# Count of each species
print("\nSpecies distribution:")
print(df['species'].value_counts())

# Mode - most common species
most_common_species = df['species'].mode()[0]
print(f"\nMost common species (mode): {most_common_species}")
print("(All three species have equal counts in this dataset)")
````

````python
# Calculate all statistics at once for all numeric columns
print("=" * 60)
print("STATISTICS FOR ALL NUMERIC COLUMNS")
print("=" * 60)

# Get only numeric columns
numeric_cols = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']

print("\nMEAN (averages):")
print(df[numeric_cols].mean().round(2))

print("\nMEDIAN (middle values):")
print(df[numeric_cols].median().round(2))

print("\nRemember: describe() gives you all of this and more!")
````

## Part 5 Summary: Statistics Cheat Sheet

```
THE THREE M'S:

    MEAN    │ Sum ÷ Count              │ df['column'].mean()
    MEDIAN  │ Middle value when sorted │ df['column'].median()
    MODE    │ Most frequent value      │ df['column'].mode()
    
    
WHEN TO USE EACH:

    • Mean: Normal data without extreme outliers
    • Median: Data with outliers (income, prices, times)
    • Mode: Categorical data or "most common" questions
    
    
RED FLAG:

    If mean and median are very different → You probably have outliers!
```

````python
# Import visualization libraries
import matplotlib.pyplot as plt  # plt is the universal convention
import seaborn as sns            # sns is the universal convention

# Set style for nice-looking plots
sns.set_style("whitegrid")  # Clean background with grid
plt.rcParams['figure.figsize'] = [10, 6]  # Default size

print("Visualization libraries loaded!")
````

## Chart Type 1: Histogram (Distribution)

```
HISTOGRAM: Shows the DISTRIBUTION of values

    "How many values fall into each range?"
    
    Example: Heights of students
    
    Count │
      20  │        ████
      15  │      ████████
      10  │    ████████████
       5  │  ████████████████
          └────────────────────────
             150  160  170  180  190 cm
    
    Most students are around 165-175 cm.
    Few are very short or very tall.
    
USE WHEN: You want to see the shape of your data distribution
```

````python
# Histogram: Distribution of sepal length
plt.figure(figsize=(10, 6))

# Create histogram with seaborn
sns.histplot(data=df, x='sepal_length', bins=20, kde=True)

# Add labels and title
plt.xlabel('Sepal Length (cm)', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.title('Distribution of Sepal Length', fontsize=14)

# Add vertical lines for mean and median
mean_val = df['sepal_length'].mean()
median_val = df['sepal_length'].median()
plt.axvline(mean_val, color='red', linestyle='--', label=f'Mean: {mean_val:.2f}')
plt.axvline(median_val, color='green', linestyle='--', label=f'Median: {median_val:.2f}')
plt.legend()

plt.tight_layout()
plt.show()

print("\nThe curve (KDE) shows the smoothed distribution.")
print("Mean and median are close → data is relatively symmetric.")
````

## Chart Type 2: Scatter Plot (Relationships)

```
SCATTER PLOT: Shows RELATIONSHIP between two variables

    "Do these two things go together?"
    
    Example: Study hours vs Exam scores
    
    Score │
      100 │                    • •
       80 │               • • •
       60 │         • • •
       40 │    • •
       20 │  •
          └───────────────────────────
              2    4    6    8    10  hours
    
    Pattern going up → Positive correlation
    (More study hours → Higher scores)
    
USE WHEN: You want to see if two numeric variables are related
```

````python
# Scatter plot: Relationship between sepal length and petal length
plt.figure(figsize=(10, 6))

# Create scatter plot colored by species
sns.scatterplot(
    data=df,
    x='sepal_length',
    y='petal_length',
    hue='species',    # Color by species
    s=100             # Marker size
)

plt.xlabel('Sepal Length (cm)', fontsize=12)
plt.ylabel('Petal Length (cm)', fontsize=12)
plt.title('Sepal Length vs Petal Length by Species', fontsize=14)

plt.tight_layout()
plt.show()

print("\nOBSERVATIONS:")
print("• Clear positive correlation: longer sepals → longer petals")
print("• Species form distinct clusters!")
print("• Setosa (purple) has smaller measurements")
print("• Virginica (green) tends to have the largest")
````

## Chart Type 3: Box Plot (Comparison)

```
BOX PLOT: Compare distributions across CATEGORIES

    Shows: Median, quartiles, and outliers
    
                    ┌───┐
           ●    ────┤   ├────●
                    └───┘
           ↑    ↑   ↑   ↑    ↑
         out  min  Q1  Q3  max  (IQR = Q3-Q1)
                    │
                  median
    
    • Box = where 50% of data lives (Q1 to Q3)
    • Line inside box = median
    • Whiskers = extend to min/max (or 1.5×IQR)
    • Dots outside = outliers
    
USE WHEN: Comparing a numeric variable across categories
```

````python
# Box plot: Compare petal length across species
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x='species',
    y='petal_length',
    palette='Set2'  # Nice color palette
)

plt.xlabel('Species', fontsize=12)
plt.ylabel('Petal Length (cm)', fontsize=12)
plt.title('Petal Length Distribution by Species', fontsize=14)

plt.tight_layout()
plt.show()

print("\nOBSERVATIONS:")
print("• Setosa has clearly smaller petals (all below 2 cm)")
print("• Versicolor is in the middle (roughly 3-5 cm)")
print("• Virginica has the longest petals (roughly 4.5-7 cm)")
print("• Versicolor has the most variability (widest box)")
````

## Chart Type 4: Bar Chart (Counts/Comparisons)

```
BAR CHART: Compare VALUES across CATEGORIES

    "How does each category compare?"
    
    Example: Sales by Region
    
    Sales │
    $100K │  ████████████████████
     $80K │  ████████████████
```

````python
# Bar Chart: Average petal length by species
plt.figure(figsize=(10, 6))

# Calculate average petal length per species
avg_by_species = df.groupby('species')['petal_length'].mean()

# Create bar chart
sns.barplot(data=df, x='species', y='petal_length', palette='viridis', errorbar=None)

plt.xlabel('Species', fontsize=12)
plt.ylabel('Average Petal Length (cm)', fontsize=12)
plt.title('Average Petal Length by Species', fontsize=14)

plt.tight_layout()
plt.show()

print("\nOBSERVATIONS:")
print("• Setosa has the shortest petals (avg ~1.5 cm)")
print("• Virginica has the longest petals (avg ~5.5 cm)")
print("• Clear difference between species - useful for classification!")
````

## Chart Type 5: Pair Plot (The Power Move)

```
PAIR PLOT: See ALL relationships at once!

    Creates a grid showing:
    • Scatter plots for every pair of numeric variables
    • Distributions on the diagonal
    • Colored by category (optional)

          (Warning: Can be slow with many columns)
```

````python
# Pair Plot: See ALL relationships at once
print("Creating pair plot (this shows ALL variable relationships)...")
print("This may take a moment...\n")

# Create pair plot - colored by species
sns.pairplot(df, hue='species', height=2.5, palette='Set2')

plt.suptitle('Iris Dataset: All Variable Relationships', y=1.02, fontsize=14)
plt.tight_layout()
plt.show()

print("\nWHAT YOU'RE SEEING:")
print("• Diagonal: Distribution of each variable (by species)")
print("• Off-diagonal: Scatter plots of variable pairs")
print("• Colors: Different species")
print("\nKEY INSIGHT: Setosa (green) is clearly separable from the others!")
````
