## The Three Operators

```
THE LOGIC OPERATORS:

    ┌──────────┬─────────────┬──────────────────────────────────────────┐
    │ Operator │ Meaning     │ English                                  │
    ├──────────┼─────────────┼──────────────────────────────────────────┤
    │    &     │ AND         │ BOTH must be true                        │
    │    |     │ OR          │ AT LEAST ONE must be true                │
    │    ~     │ NOT         │ FLIP true ↔ false                        │
    └──────────┴─────────────┴──────────────────────────────────────────┘


⚠️  THE RULE THAT WILL SAVE YOU HOURS OF DEBUGGING:

    ┌─────────────────────────────────────────────────────────────────────┐
    │                                                                     │
    │   ALWAYS wrap each condition in parentheses!                       │
    │                                                                     │
    │   ❌ WRONG:   df[df['age'] > 30 & df['city'] == 'NYC']             │
    │              → ERROR! Python evaluates in wrong order              │
    │                                                                     │
    │   ✓ RIGHT:   df[(df['age'] > 30) & (df['city'] == 'NYC')]          │
    │              → Works perfectly                                      │
    │                                                                     │
    └─────────────────────────────────────────────────────────────────────┘
```

## AND: Both Conditions Must Be True

````python
# Find July data from 1950 onward
# (year >= 1950) AND (month == 'Jul')

fifties_july = flights[(flights['year'] >= 1950) & (flights['month'] == 'Jul')]

print(f"July months from 1950+: {len(fifties_july)} rows")
fifties_july
````

## OR: At Least One Condition Must Be True

````python
# Find summer months - June OR July OR August

summer = flights[
    (flights['month'] == 'Jun') |
    (flights['month'] == 'Jul') |
    (flights['month'] == 'Aug')
]

print(f"Summer months: {len(summer)} rows")
summer.head()
````

## The Better Way: .isin()

```
THE UGLY WAY vs THE CLEAN WAY:

    ❌ UGLY (doesn't scale):
    
       (df['month'] == 'Jun') | (df['month'] == 'Jul') | (df['month'] == 'Aug')
       
       What if you need 20 values? 20 conditions, 19 pipes. Nightmare.
    
    
    ✓ CLEAN (scales beautifully):
    
       df['month'].isin(['Jun', 'Jul', 'Aug'])
       
       Need 20 values? Just add them to the list. Done.
    
    
    ┌─────────────────────────────────────────────────────────────────────┐
    │                                                                     │
    │   .isin() = "Is this value IN this list?"                          │
    │                                                                     │
    │   This is how professionals write pandas code.                     │
    │                                                                     │
    └─────────────────────────────────────────────────────────────────────┘
```

````python
# Same result, MUCH cleaner
summer_months = ['Jun', 'Jul', 'Aug', 'Sep']
summer = flights[flights['month'].isin(summer_months)]

print(f"Summer months: {len(summer)} rows")
summer.head()
````

````python
# Combine isin with other conditions
# Winter months in 1960
winter_months = ['Dec', 'Jan', 'Feb']

winter_60 = flights[
    (flights['month'].isin(winter_months)) &
    (flights['year'] == 1960)
]

print(f"Winter 1960: {len(winter_60)} rows")
winter_60
````

## Another Shortcut: .between()

Instead of:
```python
(value >= 100) & (value <= 200)
```

Use:
```python
value.between(100, 200)
```

````python
# Find moderate traffic months (200-350 passengers)
moderate = flights[flights['passengers'].between(200, 350)]

print(f"Moderate traffic: {len(moderate)} rows")
moderate.head()
````

## NOT: Inverting Conditions with ~

````python
# Everything EXCEPT summer
non_summer = flights[~flights['month'].isin(['Jun', 'Jul', 'Aug'])]

print(f"Non-summer months: {len(non_summer)} rows")
print(f"Months included: {list(non_summer['month'].unique())}")
````

## Let's Answer That Manager Question

Remember: "Summer months in the late 1950s with high traffic"

````python
# Summer months (Jun, Jul, Aug)
# Late 1950s (year >= 1957)
# High traffic (passengers > 400)

answer = flights[
    (flights['month'].isin(['Jun', 'Jul', 'Aug'])) &
    (flights['year'] >= 1957) &
    (flights['passengers'] > 400)
]

print(f"Summer months, late 50s, high traffic: {len(answer)} rows")
answer
````

## Filtering Cheat Sheet

| Need | Use | Example |
|------|-----|----------|
| Both conditions | `&` | `(A > 5) & (B < 10)` |
| Either condition | `\|` | `(A > 5) \| (B < 10)` |
| Value in list | `.isin()` | `col.isin(['x', 'y', 'z'])` |
| Numeric range | `.between()` | `col.between(10, 20)` |
| Invert/exclude | `~` | `~col.isin(['x', 'y'])` |

**Golden Rule**: Parentheses around EVERY condition!

# Part 4: loc and iloc - Precision Selection

## This Is What Separates Amateurs from Professionals

---

## The Core Difference

```
iloc vs loc: TWO WAYS TO SELECT DATA

    ┌─────────────────────────────────────────────────────────────────────┐
    │                                                                     │
    │   iloc = "Integer Location"          loc = "Label Location"        │
    │   ─────────────────────────          ─────────────────────         │
    │                                                                     │
    │   Select by POSITION                 Select by LABELS/CONDITIONS   │
    │   (0, 1, 2, 3...)                    (names, True/False)           │
    │                                                                     │
    │   "Give me row NUMBER 5"             "Give me rows WHERE age > 30" │
    │                                                                     │
    │   Think: COUNTING                    Think: DESCRIBING             │
    │                                                                     │
    └─────────────────────────────────────────────────────────────────────┘


    ANALOGY:
    
    iloc  →  "Give me the 5th person in line"
    loc   →  "Give me everyone wearing a red shirt"
```

## iloc - Selecting by Position

````python
# Single row (position 0 = first row)
flights.iloc[0]
````

````python
# Single cell: row 0, column 2
flights.iloc[0, 2]
````

````python
# Range of rows: first 5 (positions 0-4)
flights.iloc[0:5]
````

````python
# Rows AND columns: first 5 rows, first 2 columns
flights.iloc[0:5, 0:2]
````

````python
flights.iloc[5:10]
````

````python
flights.iloc[5:10:2] # adding a step of 2
````

````python
# Last 5 rows (negative indexing)
flights.iloc[-5:]
````

## loc - Selecting by Labels

````python
# Row with index 0, column named 'passengers'
flights.loc[0, 'passengers']
````

````python
# Specific rows and columns by name
flights.loc[0:4, ['year', 'month', 'passengers']]
````

````python
# CRITICAL DIFFERENCE
print(f"iloc[0:5] returns {len(flights.iloc[0:5])} rows (excludes position 5)")
print(f"loc[0:5] returns {len(flights.loc[0:5])} rows (INCLUDES index 5)")
````

## The Power Move: loc + Conditions

```
THE PATTERN THAT HANDLES 80% OF YOUR DATA WORK:

    ┌─────────────────────────────────────────────────────────────────────┐
    │                                                                     │
    │   df.loc[condition, ['column1', 'column2']]                        │
    │          ─────────  ──────────────────────                         │
    │              │               │                                      │
    │              │               └── WHICH COLUMNS to show             │
    │              │                                                      │
    │              └── WHICH ROWS to include                             │
    │                                                                     │
    │                                                                     │
    │   English: "Give me rows WHERE this is true,                       │
    │             and ONLY SHOW me these columns."                       │
    │                                                                     │
    └─────────────────────────────────────────────────────────────────────┘


    EXAMPLE:
    
    "Transactions over $10K from California"
    
    df.loc[(df['amount'] > 10000) & (df['state'] == 'CA'),
           ['customer', 'amount', 'date']]
    
    One line. Done.
```

````python
# High traffic months - only show year, month, passengers
flights.loc[flights['passengers'] > 500, ['year', 'month', 'passengers']]
````

````python
# Multiple conditions - late 50s high traffic
flights.loc[
    (flights['year'] >= 1957) & (flights['passengers'] > 450),
    ['year', 'month', 'passengers']
]
````

## loc Also Sets Values

````python
# Create a copy (always!)
df = flights.copy()

# Create a new column based on condition
df.loc[df['passengers'] > 400, 'traffic'] = 'High'
df.loc[df['passengers'] <= 400, 'traffic'] = 'Normal'

df[['year', 'month', 'passengers', 'traffic']].head(15)
````

## Quick Reference

| What you need | Use this |
|---------------|----------|
| Row by position | `iloc[5]` |
| Last 10 rows | `iloc[-10:]` |
| Rows 10-20 | `iloc[10:21]` |
| Filter + select columns | `loc[condition, ['col1', 'col2']]` |
| Set values by condition | `loc[condition, 'col'] = value` |

**Rule of thumb**: iloc for positions, loc for everything else.

````python
# Load titanic - real data with real problems
titanic = sns.load_dataset('titanic')

print(f"Loaded {len(titanic)} passenger records")
titanic.head()
````

````python
# The damage report
print("Missing values:")
print(titanic.isna().sum())
print(f"\nTotal rows: {len(titanic)}")
````

## Option 1: Just Drop Missing Rows

````python
# The naive approach
clean = titanic.dropna()

print(f"Before: {len(titanic)} rows")
print(f"After: {len(clean)} rows")
print(f"Lost: {len(titanic) - len(clean)} rows ({(len(titanic)-len(clean))/len(titanic)*100:.0f}%)")
````

## Option 2: Fill Missing Values

````python
# For numbers: fill with median (not mean - outliers!)
df = titanic.copy()
median_age = df['age'].median()

print(f"Median age: {median_age}")

df['age'] = df['age'].fillna(median_age)
print(f"Missing ages now: {df['age'].isna().sum()}")
````

````python
# For categories: fill with mode (most common)
df = titanic.copy()
most_common = df['embarked'].mode()[0]

print(f"Most common port: {most_common}")

df['embarked'] = df['embarked'].fillna(most_common)
print(f"Missing embarked now: {df['embarked'].isna().sum()}")
````

## Option 3: Drop Hopeless Columns

````python
# Deck is 77% missing - just drop it
df = titanic.drop(columns=['deck'])

print(f"Dropped 'deck' column")
print(f"Columns: {len(titanic.columns)} -> {len(df.columns)}")
````

## The Real-World Recipe

````python
# What professionals actually do
df = titanic.copy()

# 1. Drop columns that are >50% missing
df = df.drop(columns=['deck'])

# 2. Fill numbers with median
df['age'] = df['age'].fillna(df['age'].median())

# 3. Fill categories with mode
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])
df['embark_town'] = df['embark_town'].fillna(df['embark_town'].mode()[0])

print("Remaining missing values:")
print(df.isna().sum().sum())
print(f"\nRows kept: {len(df)} / {len(titanic)} = 100%")
````

````python
# SOLUTION Q1: Which day has highest average tip?
# We need groupby for proper averages (Day 3 topic), but we can approximate:

print("Average tip by day:")
for day in tips['day'].unique():
    avg = tips[tips['day'] == day]['tip'].mean()
    print(f"  {day}: ${avg:.2f}")

# Or use value_counts to see which days are busiest
print("\nNumber of transactions by day:")
print(tips['day'].value_counts())
````

````python
# SOLUTION Q2: Dinner parties 4+ people, tip > $5
# Multiple conditions + loc for specific columns

big_dinner_tippers = tips.loc[
    (tips['time'] == 'Dinner') &
    (tips['size'] >= 4) &
    (tips['tip'] > 5),
    ['day', 'time', 'size', 'total_bill', 'tip']
]

print(f"Found {len(big_dinner_tippers)} dinner parties of 4+ with tips over $5:")
big_dinner_tippers
````

````python
# SOLUTION Q3: Top 10 tippers - who are they?
top_tippers = tips.nlargest(10, 'tip')

print("Top 10 tippers:")
print(top_tippers)

print("\n" + "="*50)
print("PATTERNS IN TOP TIPPERS:")
print("="*50)
print(f"\nTime of day: {top_tippers['time'].value_counts().to_dict()}")
print(f"Day of week: {top_tippers['day'].value_counts().to_dict()}")
print(f"Smoker: {top_tippers['smoker'].value_counts().to_dict()}")
print(f"Average party size: {top_tippers['size'].mean():.1f}")
print(f"Average total bill: ${top_tippers['total_bill'].mean():.2f}")
````

````python
# SOLUTION Q4: Clean missing tip data
tips_clean = tips_messy.copy()

# Check before
print(f"Missing tips before: {tips_clean['tip'].isna().sum()}")

# Fill with median (tip is numeric)
median_tip = tips_clean['tip'].median()
print(f"Median tip: ${median_tip:.2f}")

tips_clean['tip'] = tips_clean['tip'].fillna(median_tip)

# Verify
print(f"Missing tips after: {tips_clean['tip'].isna().sum()}")
print(f"\nData cleaned! All {len(tips_clean)} rows preserved.")
````
