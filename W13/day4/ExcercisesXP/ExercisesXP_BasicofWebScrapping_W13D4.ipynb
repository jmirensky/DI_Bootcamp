🌟 Exercise 3 : Extracting Headers from Wikipedia’s Main Page

```python
url = "https://www.wikipedia.org/"

headers = HEADERS

response = requests.get(url, headers=headers)
response.raise_for_status()
soup = BeautifulSoup(response.text, "html.parser")

# Search all headers h1 to h6
header_tags = soup.find_all(["h1", "h2", "h3", "h4", "h5", "h6"])

for h in header_tags:
    print(h.name, ":", h.get_text(" ", strip=True))
```

🌟 Exercise 5 : Analyzing US-CERT Security Alerts

Write a Python program to get the number of security alerts issued by US-CERT in the current year.
Source --> https://www.cisa.gov/news-events/cybersecurity-advisories?f%5B0%5D=advisory_type%3A93

```python
base_url = "https://www.cisa.gov/news-events/cybersecurity-advisories"
params = "?f%5B0%5D=advisory_type%3A93&page="

headers = HEADERS

current_year = str(datetime.now().year)

count = 0
page = 0

while True:
    url = base_url + params + str(page)
    response = requests.get(url, headers=headers)
    response.raise_for_status()
    soup = BeautifulSoup(response.text, "html.parser")

    advisories = soup.find_all("article")

    # If there are no results → we're done
    if not advisories:
        break

    stop = False

    for adv in advisories:
        text = adv.get_text()

        if current_year in text:
            count += 1
        else:
            # if we find another year → we stop
            stop = True
            break

    if stop:
        break

    page += 1

print("Total alerts en", current_year, ":", count)
```

🌟 Exercise 6 : Scraping Movie Details

Write a Python program to get movie name, year and a brief summary of the top 10 random movies from the IMBD website.

```python
# IMDb uses dynamic content and anti-scraping measures, so it is not accessible with basic requests.
# An alternative public source was used instead.

# 🌟 Exercise 6 : Scraping Movie Details

# IMDb uses dynamic content and anti-scraping measures, so it is not accessible with basic requests.
# An alternative public source was used instead.

url = "https://en.wikipedia.org/wiki/List_of_American_films_of_2021"
headers = HEADERS

try:
    res = requests.get(url, headers=headers)
    res.raise_for_status()
except requests.exceptions.RequestException as e:
    print(f"Error al obtener la lista de películas: {e}")
    res = None

if res:
    soup = BeautifulSoup(res.text, "html.parser")

    tables = soup.find_all("table", class_="wikitable")

    movies = []

    for table in tables:
        rows = table.find_all("tr")[1:]
        for row in rows:
            cols = row.find_all("td")
            if cols:
                a = cols[0].find("a")
                if a:
                    title = a.get_text(strip=True)
                    link = "https://en.wikipedia.org" + a["href"]
                    movies.append((title, link))

    sample = random.sample(movies, 10)

    for title, link in sample:
        try:
            r = requests.get(link, headers=headers)
            r.raise_for_status()
        except requests.exceptions.RequestException as e:
            print(f"Error al obtener {title}: {e}")
            continue

        s = BeautifulSoup(r.text, "html.parser")

        # find the first REAL (not empty) paragraph
        summary = "N/A"
        for p in s.find_all("p"):
            text = p.get_text(strip=True)
            if len(text) > 40:  # basic quality filter
                summary = text
                break

        print("title: ", title)
        print("summary: ", summary)
        print("-" * 50)
```
