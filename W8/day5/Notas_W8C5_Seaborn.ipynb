## Chart Selection Quick Reference

```
    ┌──────────────────────────────────────────────────────────────────────┐
    │ YOUR QUESTION                     │ USE THIS CHART                   │
    ├───────────────────────────────────┼──────────────────────────────────┤
    │ "How is X distributed?"           │ Histogram, KDE                   │
    │ "How many in each category?"      │ Count plot, Bar plot             │
    │ "How do groups compare on Y?"     │ Bar plot (means), Box plot       │
    │ "How do X and Y relate?"          │ Scatter plot                     │
    │ "How does Y change over time?"    │ Line plot                        │
    │ "How do distributions differ?"    │ Box plot, Violin plot            │
    │ "What are ALL the relationships?" │ Pair plot                        │
    │ "What correlates with what?"      │ Heatmap                          │
    └───────────────────────────────────┴──────────────────────────────────┘

    THE KEY INSIGHT:
    Start with the QUESTION, not the chart.
    "I want to make a bar chart" → WRONG approach
    "I want to compare average weight across species" → RIGHT approach
        → Then you realize: bar chart or box plot would work
```

## Key Takeaways

```
TODAY'S LESSONS:

    1. QUESTION FIRST, CHART SECOND
       "How is this distributed?" → Histogram
       "How do groups compare?" → Bar/Box plot
       "How do X and Y relate?" → Scatter plot

    2. WEIRD DISTRIBUTIONS = MIXED GROUPS
       When you see bimodal or strange shapes,
       add hue to reveal hidden groups.

    3. AGGREGATE DATA CAN LIE (Simpson's Paradox)
       Always ask: "Am I mixing groups that should
       be analyzed separately?"

    4. PAIR PLOTS = EXPLORATORY SUPERPOWER
       See all relationships at once, then dive deeper.

    5. DASHBOARDS TELL COMPLETE STORIES
       Real analysis = multiple visualizations working together.
```

## Quick Reference: Today's Patterns

```python
# HISTOGRAM - Distribution of continuous variable
sns.histplot(data=df, x='numeric_col', kde=True)
sns.histplot(data=df, x='numeric_col', hue='category', alpha=0.6)

# COUNT PLOT - How many in each category
sns.countplot(data=df, x='category')
sns.countplot(data=df, x='category', hue='subcategory')

# BAR PLOT - Compare means (with confidence intervals)
sns.barplot(data=df, x='category', y='value')
sns.barplot(data=df, x='category', y='value', hue='subcategory')

# SCATTER PLOT - Relationship between two variables
sns.scatterplot(data=df, x='var1', y='var2', hue='category')
sns.scatterplot(data=df, x='var1', y='var2', hue='category', size='numeric', style='sex')

# BOX PLOT - Compare distributions
sns.boxplot(data=df, x='category', y='value')
sns.boxplot(data=df, x='category', y='value', hue='subcategory')

# PAIR PLOT - All relationships at once
sns.pairplot(df, hue='category', diag_kind='kde')
sns.pairplot(df, hue='category', corner=True)  # Less redundant

# DASHBOARD
fig, axes = plt.subplots(2, 2, figsize=(14, 12))
# ... create plots on axes[0,0], axes[0,1], etc.
plt.suptitle('Dashboard Title', fontweight='bold')
plt.tight_layout()
plt.savefig('dashboard.png', dpi=300, bbox_inches='tight')
```

📈 Distribution Plots (Distribuciones)



1️⃣ Histogramas (histplot)

Muestran la frecuencia de una variable.

Se pueden ajustar bins, color, estilo.

Útiles para ver forma, asimetría y dispersión.

2️⃣ KDE (kdeplot)

Curva suave que estima la densidad de probabilidad.

Más elegante que el histograma.

Puede ajustarse el bandwidth.

fill=True rellena el área bajo la curva.

3️⃣ Univariado vs Bivariado

Univariado → una sola variable.

Bivariado → relación entre dos variables (jointplot).

🎨 Personalización (NO LO VIMOS ?)

set_style() para cambiar el fondo.

set_palette() para colores.

````python
# ============================================
# RUN THIS CELL FIRST - All imports
# ============================================

import pandas as pd   #for data  manipulation
import matplotlib.pyplot as plt    #for general plotting
import matplotlib.ticker as mtick
import seaborn as sns    #for advanced visualization
import numpy as np    #for numeric operations
from scipy import stats   #for custom KDE calculation

print("All libraries loaded")
````

````python
#formatos predefinidos

#todos los gráficos de Matplotlib/Seaborn aparezcan dentro del notebook, justo debajo de la celda donde los generás.
%matplotlib inline

#muestra hasta 100 filas cuando haces df.head() o df.
pd.set_option('display.max_rows', 100)

#muestra hasta 20 columnas completas.
pd.set_option('display.max_columns', 20)

# -----------------------------
#  Global context and style
# -----------------------------

sns.set_context("paper", font_scale=0.8)  # más chico y legible, reduce todo proporcionalmente

sns.set_style("whitegrid")

sns.set_palette("Set3")
    # Set1, Set2, Set3 → colores fuertes y distintos
    # Pastel1, Pastel2 → tonos más suaves
    # Dark2, Accent → tonos más corporativos, sobrios

    # Paletas secuenciales (para gradientes, números)
    # Para variables con orden, como revenue:
    # sns.set_palette("Blues")       # azul claro → azul oscuro
    # sns.set_palette("Greens")
    # sns.set_palette("Reds")
    # sns.set_palette("Purples")
    # sns.set_palette("Oranges")


# Default size of all figures
plt.rcParams["figure.figsize"] = (10,5)  # ancho 6, alto 4 pulgadas
plt.rcParams["axes.titlesize"] = 12
plt.rcParams["axes.titleweight"] = "bold"
plt.rcParams["axes.labelsize"] = 10
plt.rcParams["xtick.labelsize"] = 8
plt.rcParams["ytick.labelsize"] = 8
plt.rcParams["legend.fontsize"] = 9
plt.rcParams["lines.linewidth"] = 2
plt.rcParams["lines.markersize"] = 4
plt.rcParams["figure.dpi"] = 100   # resolución decente, no gigante


# -----------------------------
# Helper function to apply formatting to any char
# -----------------------------
def format_plot(ax, show_numbers=False):
    """
    Aplica formato consistente a cualquier gráfico:
    - Eje Y con punto de mil
    - Bordes eliminados (sns.despine)
    - Opcional: números centrados sobre barras
    """
    # Eje Y con separador de miles
    ax.yaxis.set_major_formatter(
        mtick.FuncFormatter(lambda val, _: f"{val:,.0f}".replace(",", "."))
    )

    # Quitar bordes
    sns.despine(ax=ax)

    # Si es un barplot, mostrar números encima de cada barra
    if show_numbers:
        for p in ax.patches:
            height = p.get_height()
            ax.text(
                x=p.get_x() + p.get_width()/2,
                y=height + 0.01*height,
                s=f"{height:,.0f}".replace(",", "."),
                ha='center',
                va='bottom',
                fontsize=8
            )

    plt.tight_layout()
    return ax

````

````python
# 1. Load and Inspect Data:

url = 'https://raw.githubusercontent.com/intern2grow/sales-data-analysis/refs/heads/main/sales_data.csv'

df_raw = pd.read_csv(url)

# head - Preview first rows
print("\nSales - first 5 rows:")
print(df_raw.head(5))

#shape - how big is the data?
print(f"\nSize: {df_raw.shape[0]} rows x {df_raw.shape[1]} columns" )

# info() -  what's missing? what data types?
print("\nSales - nulls and data type:")
print(df_raw.info())


# describe() - statistics for numeric columns
print(f'\nSummary statistics: {df_raw.describe()}')
````

````python
# Count duplicates:
print(f'\nDuplicate rows: {df_raw.duplicated().sum()}')

# Filter duplicate rows
duplicates = df_raw[df_raw.duplicated(keep=False)]  # keep=False marca todas las ocurrencias
print(duplicates)

print(f"\nSize Before: {df_raw.shape[0]} rows")

df_noduplicates = df_raw.drop_duplicates()

print(f"\nSize After: {df_noduplicates.shape[0]} rows")
````

````python
# 2. Data Preprocessing

# Handle missing values
print(f'\nMissing values:{df_noduplicates.isna().sum()}')

missing_counts = df_noduplicates.isnull().sum()
missing_pct = ((missing_counts / len(df_noduplicates)) * 100).round(1)

missing_df_noduplicates = pd.DataFrame(
    {
        'Missing Count': missing_counts,
        'Missing %': missing_pct
    }
)

missing_df_noduplicates[missing_df_noduplicates['Missing Count'] > 0].sort_values(by='Missing %', ascending=False)
````

````python
# Removing NA, there are too few of them, and they don't carry much weight.

df = df_noduplicates.dropna()

print(f"\nSize Before dropping NA: {df_noduplicates.shape[0]} rows")

print(f"\nSize After: {df.shape[0]} rows")

print(f'\nMissing values:{df.isna().sum()}')
````
